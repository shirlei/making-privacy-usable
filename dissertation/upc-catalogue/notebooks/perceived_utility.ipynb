```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots


# === Data  from the sheet 'perception_en' in the spreadsheet ../data/evaluation1-post-activity-questionnaire.xlsx ===
df_inst1 = {
    "Question": [
        "The guidelines facilitated the proposal of improvements to the analysed system (M2.1)",
        "It was easy to identify guidelines that matched the identified problems (M2.2)",
        "The guidelines are adequate for real-world digital system contexts (M2.3)",
        "The guidelines helped me understand legal data protection principles and rights (M2.4)",
        "I now feel more prepared to develop digital systems with data protection in mind (M2.5)"
    ],
    "1": [0, 0, 0, 0, 0],
    "2": [0, 2, 0, 0, 2],
    "3": [1, 8, 1, 1, 3],
    "4": [6, 6, 5, 9, 11],
    "5": [25, 16, 26, 22, 16],
    "N": [32, 32, 32, 32, 32]
}

# === Data  from the sheet 'perception_en' in the spreadsheet ../data/evaluation2-post-activity-questionnaire.xlsx ===
df_inst2 = {
    "Question": [
        "The guidelines facilitated the proposal of improvements to the analysed system (M2.1)",
        "It was easy to identify guidelines that matched the identified problems (M2.2)",
        "The guidelines are adequate for real-world digital system contexts (M2.3)",
        "The guidelines helped me understand legal data protection principles and rights (M2.4)",
        "I now feel more prepared to develop digital systems with data protection in mind (M2.5)"
    ],
    "1": [0, 1, 0, 0, 0],
    "2": [0, 3, 1, 1, 1],
    "3": [5, 7, 4, 3, 2],
    "4": [8, 6, 6, 8, 13],
    "5": [10, 6, 12, 11, 7],
    "N": [23, 23, 23, 23, 23]
}


# === Data  from the sheet 'perception_en' in the spreadsheet ../data/evaluation3-post-activity-questionnaire.xlsx ===
df_inst3 = {
    "Question": [
        "The guidelines facilitated the proposal of improvements to the analysed system (M2.1)",
        "It was easy to identify guidelines that matched the identified problems (M2.2)",
        "The guidelines are adequate for real-world digital system contexts (M2.3)",
        "The guidelines helped me understand legal data protection principles and rights (M2.4)",
        "I now feel more prepared to develop digital systems with data protection in mind (M2.5)"
    ],
    "1": [0, 0, 0, 0, 0],
    "2": [0, 0, 0, 0, 1],
    "3": [0, 2, 1, 0, 1],
    "4": [5, 11, 3, 6, 6],
    "5": [14, 6, 15, 13, 11],
    "N": [19, 19, 19, 19, 19]
}

df1 = pd.DataFrame(df_inst1)
df2 = pd.DataFrame(df_inst2)
df3 = pd.DataFrame(df_inst3)
response_order = ["1", "2", "3", "4", "5"]

# Use compact y-axis labels and move full questions to a separate legend area.
short_labels = df1["Question"].str.extract(r"\(([^)]+)\)\s*$")[0].fillna(df1["Question"])
# question_legend_lines = [
#     f"{code}: {question.split(' (')[0].replace('<br>', ' ')}"
#     for code, question in zip(short_labels, df1["Question"])
# ]
# question_legend_text = "<br>".join(question_legend_lines)
question_legend_lines = [
    f"{code}: {question.split(' (')[0].replace('<br>', ' ')}"
    for code, question in zip(reversed(short_labels), reversed(df1["Question"]))
]
question_legend_text = "<br>".join(question_legend_lines)

colors = ['#deebf7', '#bcbddc', '#756bb1', '#3182bd', '#08519c']

fig = make_subplots(
    rows=1,
    cols=3,
    shared_yaxes=True,
    horizontal_spacing=0.03,
    subplot_titles=["1st application", "2nd application", "3rd application"]
)


def add_likert_bars(fig, df, col, show_legend):
    for i, response in enumerate(response_order):
        counts = df[response]
        percentages = (counts / df["N"]) * 100
        labels = [f"{p:.1f}%" if p > 0 else "" for p in percentages]

        fig.add_trace(
            go.Bar(
                y=short_labels,
                x=percentages,
                orientation='h',
                name=response,
                marker_color=colors[i],
                text=labels,
                textposition='inside',
                insidetextanchor='middle',
                showlegend=show_legend
            ),
            row=1,
            col=col
        )

add_likert_bars(fig, df1, col=1, show_legend=False)
add_likert_bars(fig, df2, col=2, show_legend=False)
add_likert_bars(fig, df3, col=3, show_legend=True)


# Show short labels only in the first subplot.
fig.update_yaxes(showticklabels=True, row=1, col=1)
fig.update_yaxes(showticklabels=False, row=1, col=2)
fig.update_yaxes(showticklabels=False, row=1, col=3)
fig.update_layout(
    barmode='stack',
    xaxis_title="Percentage of responses (%)",
    legend_title="Likert response",
    font=dict(size=14),
    plot_bgcolor='white',
    height=900,
    width=1400,
    legend=dict(
        orientation="h",
        x=0.25,          # legenda à esquerda
        xanchor="center",
        y=-0.12,
        yanchor="top",
        traceorder="normal",
    ),
    margin=dict(l=40, r=40, t=80, b=200)
)

fig.add_annotation(
    x=0.52,             # à direita da legenda
    y=-0.22,            # mesma altura da legenda
    xref="paper",
    yref="paper",
    showarrow=False,
    align="left",
    xanchor="left",
    text=f"<b>Questionnaire statement legend:</b><br>{question_legend_text}",
    font=dict(size=11)
)

fig.show()
fig.write_image("../images/perception_upc_guidelines.png", scale=3)
```
