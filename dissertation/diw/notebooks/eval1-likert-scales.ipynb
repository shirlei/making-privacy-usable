```python
import pandas as pd
import plotly.graph_objects as go

df_inst1 = {
    "Question": [
        "M1.1: I felt confident about who was <br>requesting my data", #b2
        "M1.3: It was clear to me which data were <br>being requested", #b3.a
        "M1.4: I was able to distinguish which data were <br>required and which were optional", #b3.c
        "M1.6: I understood why each piece of data was <br>being requested", #b3.b
        "M1.8: I felt that I had control over which data <br>I was sharing (M1.8)", #b6.a
        "M1.9: The Minha Carteira app encouraged <br>me to think before sharing my data", #b6.b
        "M1.10: I was able to recognise when I was <br>sharing more data than necessary" #b6.c
    ],
    "1": [0, 0, 0, 0, 0, 1, 3],
    "2": [0, 0, 1, 1, 1, 0, 0],
    "3": [2, 2, 1, 1, 2, 0, 4],
    "4": [5, 1, 1, 2, 6, 7, 4],
    "5": [12, 16, 16, 15, 10, 11, 8],
    "N": [19, 19, 19, 19, 19, 19, 19]
}

df1 = pd.DataFrame(df_inst1)
response_order = ["1", "2", "3", "4", "5"]
colors = ['#deebf7', '#bcbddc', '#756bb1', '#3182bd', '#08519c']

fig = go.Figure()

for i, response in enumerate(response_order):
    counts = df1[response]
    percentages = (counts / df1["N"]) * 100
    labels = [f"{p:.1f}%" if p > 0 else "" for p in percentages]

    fig.add_trace(
        go.Bar(
            y=df1["Question"],
            x=percentages,
            orientation='h',
            name=response,
            marker_color=colors[i],
            text=labels,
            textposition='inside',
            insidetextanchor='middle'
        )
    )

fig.update_layout(
    barmode='stack',
    xaxis_title="Percentage of responses (%)",
    legend_title="Likert response",
    font=dict(size=14),
    plot_bgcolor='white',
    height=550,
    width=1200,
    margin=dict(l=40, r=40, t=40, b=40)
)
fig.update_layout(
    legend=dict(
        traceorder="normal",
    )
)
fig.show()
fig.write_image("../images/prototype-likert_inst1.png", scale=3)
```
