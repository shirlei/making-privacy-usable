```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots



# === Data  from the sheet 'perception_en' in the spreadsheet ../data/evaluation1-post-activity-questionnaire.xlsx ===
df_inst1 = {
    "Question": [
        "The guidelines facilitated the proposal of <br> improvements to the analysed system (M2.1)",
        "It was easy to identify guidelines that <br> matched the identified problems (M2.2)",
        "The guidelines are adequate for real-world <br> digital system contexts (M2.3)",
        "The guidelines helped me understand legal <br> data protection principles and rights (M2.4)",
        "I now feel more prepared to develop digital <br> systems with data protection in mind (M2.5)"
    ],
    "1": [0, 0, 0, 0, 0],
    "2": [0, 2, 0, 0, 2],
    "3": [1, 8, 1, 1, 3],
    "4": [6, 6, 5, 9, 11],
    "5": [25, 16, 26, 22, 16],
    "N": [32, 32, 32, 32, 32]
}

# === Data  from the sheet 'perception_en' in the spreadsheet ../data/evaluation2-post-activity-questionnaire.xlsx ===
df_inst2 = {
    "Question": [
        "The guidelines facilitated the proposal of <br> improvements to the analysed system (M2.1)",
        "It was easy to identify guidelines that <br> matched the identified problems (M2.2)",
        "The guidelines are adequate for real-world <br> digital system contexts (M2.3)",
        "The guidelines helped me understand legal <br> data protection principles and rights (M2.4)",
        "I now feel more prepared to develop digital <br> systems with data protection in mind (M2.5)"
    ],
    "1": [0, 1, 0, 0, 0],
    "2": [0, 3, 1, 1, 1],
    "3": [5, 7, 4, 3, 2],
    "4": [8, 6, 6, 8, 13],
    "5": [10, 6, 12, 11, 7],
    "N": [23, 23, 23, 23, 23]
}

df1 = pd.DataFrame(df_inst1)
df2 = pd.DataFrame(df_inst2)

response_order = ["1", "2", "3", "4", "5"]

colors = ['#deebf7', '#bcbddc', '#756bb1', '#3182bd', '#08519c']

fig = make_subplots(
    rows=1,
    cols=2,
    shared_yaxes=True,
    horizontal_spacing=0.03,
    subplot_titles=["First application", "Second application"]
)


def add_likert_bars(fig, df, col, show_legend):
    for i, response in enumerate(response_order):
        counts = df[response]
        percentages = (counts / df["N"]) * 100
        labels = [f"{p:.1f}%" if p > 0 else "" for p in percentages]

        fig.add_trace(
            go.Bar(
                y=df["Question"],
                x=percentages,
                orientation='h',
                name=response,
                marker_color=colors[i],
                text=labels,
                textposition='inside',
                insidetextanchor='middle',
                showlegend=show_legend
            ),
            row=1,
            col=col
        )

add_likert_bars(fig, df1, col=1, show_legend=False)
add_likert_bars(fig, df2, col=2, show_legend=True)


# show questions only in left subplot
fig.update_yaxes(showticklabels=True, row=1, col=1)
fig.update_yaxes(showticklabels=False, row=1, col=2)

fig.update_layout(
    barmode='stack',
    xaxis_title="Percentage of responses (%)",
    legend_title="Likert response",
    font=dict(size=14),
    plot_bgcolor='white',
    height=550,
    width=1400,
    margin=dict(l=40, r=40, t=80, b=40)
)

fig.show()
fig.write_image("../images/figure13.png", scale=3)
```
