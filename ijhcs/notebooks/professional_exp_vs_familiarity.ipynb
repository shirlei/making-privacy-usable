```python
import pandas as pd
import plotly.express as px

# === Data translated from the sheet 'familiarity-vs-work-experience' in the spreadsheet ../data/evaluation1-post-activity-questionnaire.xlsx and familiarity-vs-work-experience from ../data/evaluation2-post-activity-questionnaire.xlsx===
data = {
    "Experience": [
        "No experience", 
        "< 1 year", 
        "1-2 years", 
        "2-5 years", 
        "> 5 years", 
        "Other fields"
    ],
    "Never had contact with the topic ": [3, 0, 1, 1, 0, 0],
    "Heard of it, never studied ": [5, 8, 7, 8, 1, 1],
    "Studied formally, never applied ": [0, 2, 7, 5, 1, 0],
    "Applied in system development ": [0, 0, 2, 2, 0, 0],
    "Data protection-related role " : [0, 0, 0, 1, 0, 0],
}

df = pd.DataFrame(data)

df_long = df.melt(id_vars="Experience", 
                  var_name="Familiarity", 
                  value_name="Count")

totals = df_long.groupby("Familiarity")["Count"].transform("sum")

df_long["Percentage"] = (df_long["Count"] / totals) * 100

df_long["Label"] = df_long.apply(
    lambda row: f"{row['Count']} ({row['Percentage']:.0f}%)" if row['Count'] > 0 else "",
    axis=1
)

# Personalised colours
colours = ['#deebf7','#9ecae1', '#bcbddc', '#756bb1', '#3182bd', '#08519c']

fig = px.bar(
    df_long,
    y="Familiarity",
    x="Count",
    color="Experience",
    orientation='h',
    barmode="group",
    text=df_long["Label"],
    title="",
    color_discrete_sequence=colours,
)

# Layout
fig.update_layout(
    yaxis_title="",
    xaxis_title="Number of Participants",
    font=dict(size=14),
    bargap=0.15,
    legend_title="Professional Experience",
    height=600,
    width=1400,
    plot_bgcolor='rgba(0,0,0,0)',
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.05,
        xanchor="center",
        x=0.2
    )
)


fig.update_traces(
    textfont_size=14,
    textposition="outside",
    textangle=0,
    insidetextanchor='middle'
)

fig.update_layout(
    margin=dict(l=100, r=50, t=80, b=80),
    xaxis=dict(range=[0, df_long['Count'].max() * 1.2]),
)

fig.write_image("../images/figure8.png", scale=3, width=900, height=600)
fig.show()
```
